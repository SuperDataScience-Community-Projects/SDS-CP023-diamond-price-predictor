- **depth**: Total depth percentage = (z / mean(x, y)) × 100
- **table**: Width of the top facet relative to the diamond’s width
- **x, y, z**: Physical dimensions of the diamond (length, width, depth in mm)
- **price** *(target)*: The market price of the diamond in USD

### 📁 3.1.1 **Diamond Color**
____

The `color` attribute represents how **colorless** a diamond is. The **Gemological Institute of America (GIA)** grades diamonds on a scale from **D (colorless)** to **J (noticeable color tint)**:

> **D (Best) → J (Least Desirable)**

- **D–F**: Colorless — extremely rare and valuable  
- **G–H**: Near-colorless — slight tint, high quality  
- **I–J**: Noticeable tint — more affordable

Color becomes more noticeable in larger diamonds. Buyers often prefer diamonds that appear white, even if they fall slightly below the top of the scale.

### 📁 3.1.2 **Diamond Cut**
___

The `cut` attribute refers to how well a diamond has been **shaped and faceted**, which directly influences its **brilliance, fire, and sparkle**. A well-cut diamond reflects light more effectively, enhancing its overall appearance and value.

Cut is assessed and graded based on proportions, symmetry, and polish. The standard grades, in **ascending order of quality**, are:

> **Fair < Good < Very Good < Premium < Ideal**

- **Fair**: Lowest cut quality; less brilliance and sparkle  
- **Good**: Acceptable quality but less optimal reflection  
- **Very Good**: High quality, reflecting light well  
- **Premium**: Excellent brilliance; close to Ideal  
- **Ideal**: Highest cut quality, offering maximum sparkle and fire

Cut is considered one of the most important characteristics influencing price because even a diamond with top color and clarity can appear dull if poorly cut.

### 📁 3.1.3 **Diamond Clarity**
___

The `clarity` attribute indicates the presence of **inclusions (internal flaws)** and **blemishes (surface imperfections)**. Clarity is graded based on visibility of flaws under 10x magnification.

The standard clarity grades, from **least to most desirable**, are:

> **I1 < SI2 < SI1 < VS2 < VS1 < VVS2 < VVS1 < IF**

- **I1 (Included)**: Visible inclusions affecting brilliance  
- **SI2, SI1 (Slightly Included)**: Inclusions visible under magnification  
- **VS2, VS1 (Very Slightly Included)**: Minor inclusions, hard to detect  
- **VVS2, VVS1 (Very Very Slightly Included)**: Minute inclusions, nearly invisible  
- **IF (Internally Flawless)**: No internal inclusions; highest clarity grade

While clarity affects value, many flaws are invisible to the naked eye, making mid-tier grades popular for their price-to-quality ratio.

## 🔍 3.3.0 Chi-Square Test of Independence: Cut × Clarity × Color
____

In this section, we apply the **Chi-Square Test of Independence** to explore whether key categorical features — **cut**, **clarity**, and **color** — are statistically dependent on one another.    

Identifying significant associations among these variables helps uncover patterns that influence diamond grading and pricing. These insights can guide strategic inventory planning, customer education, and feature engineering for predictive modeling.


````python
def chi_square_test(cat_var1, cat_var2, df, alpha=0.05):
    print(f"\n🔍 Testing Association Between: '{cat_var1}' and '{cat_var2}'")
    print("-------------------------------------------------")
    
    # Create contingency table
    table = pd.crosstab(df[cat_var1], df[cat_var2])
    print("\nContingency Table:")
    print(table)
    # Chi-square test
    chi2, p, dof, expected = chi2_contingency(table)
    
    print(f"\nChi-square Statistic: {chi2:.4f}")
    print(f"Degrees of Freedom: {dof}")
    print(f"P-value: {p:.4f}")
 
    # Interpretation
    if p < alpha:
        print(f"Significant association between '{cat_var1}' and '{cat_var2}' (p < {alpha})")
    else:
        print(f" No significant association between '{cat_var1}' and '{cat_var2}' (p >= {alpha})")

# Apply tests
chi_square_test('cut', 'clarity', df)
chi_square_test('cut', 'color', df)
chi_square_test('clarity', 'color', df)
````

*  Statistical testing reveals a significant association between <font color="red">cut</font> and <font color="red">clarity</font>, suggesting that higher-quality cuts are more frequently paired with premium clarity grades. For instance, diamonds with Ideal cuts are heavily skewed towards VS1–IF clarity, making them appealing to high-end buyers.   

* Similarly, there’s a strong association between <font color="red">cut</font> and <font color="red">color</font>, reinforcing that diamonds with Ideal and Very Good cuts are often found in the D–F color range. This combination offers both brilliance and visual purity, making it a powerful upselling point.

The K-Best selection method ranks the features based on their predictive power and selects the top k features that have the highest impact on the model's performance. We will use this method to identify the most important features for predicting diamond prices.

To use the K-Best selection method, we need to encode the categorical columns using one-hot encoding. This will convert the categorical columns into numerical columns that can be used in the feature selection process.

To facilitate the feature selection process, we build pipelines that include the one-hot encoding and K-Best selection steps. This will help us to streamline the feature selection process and identify the most important features for predicting diamond prices.


````python
# Split features and target
X = df.drop(columns='price')
y = df['price']

# Identify column types
numerical_cols = X.select_dtypes(include=['float64', 'int64']).columns.tolist()
categorical_cols = X.select_dtypes(include='object').columns.tolist()

# Create ColumnTransformer for preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(drop='first'), categorical_cols)
    ],
    remainder='passthrough',
    verbose_feature_names_out=False
)

# Pipeline for One-Hot Encoding + K-Best selection
k = 5
pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('feature_selection', SelectKBest(score_func=f_regression, k=k))
])

# Fit the pipeline
pipeline.fit(X, y)
````

````python
# Get feature names after one-hot encoding + passthrough
ohe_feature_names = pipeline.named_steps['preprocessor'].get_feature_names_out()
all_feature_names = np.array(ohe_feature_names)

# Mask of selected features
mask = pipeline.named_steps['feature_selection'].get_support()

# Selected top K feature names
selected_features = all_feature_names[mask]
print("Top K Selected Features:")
print(selected_features)
````

## 🔍 4.7 Check for Multicollinearity on Selected Features
___

Before proceeding with model training, it’s essential to assess the selected features for multicollinearity—a condition where two or more predictor variables are highly correlated. Multicollinearity can inflate the variance of coefficient estimates, making them unstable and difficult to interpret. This can ultimately lead to overfitting, reduce model interpretability, and distort the impact of individual features on the target variable.

In line with the features recommended by the SelectKBest method, I will evaluate the top 5 predictors selected for diamond price estimation. To ensure these features contribute independently and reliably to the model, I’ll use two key diagnostics:

1. Correlation Matrix – to identify pairwise relationships among features.
2. Variance Inflation Factor (VIF) – to quantify the degree of multicollinearity in the dataset.


This process is crucial for building a robust, interpretable model that generalizes well to unseen data.


````python
# Selected features
selected_features = ['carat', 'clarity', 'x_factor', 'y_factor', 'z_factor']
df_selected = df[selected_features].copy()

# Encode 'clarity' if it's categorical
if df_selected['clarity'].dtype == 'object':
    le = LabelEncoder()
    df_selected['clarity'] = le.fit_transform(df_selected['clarity'])

# Compute correlation matrix
corr_matrix = df_selected.corr()

# Create mask for upper triangle
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

# Plot lower triangle heatmap
plt.figure(figsize=(15, 6))
sns.heatmap(
    corr_matrix,
    mask=mask,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5,
    vmin=-1,
    vmax=1
)
plt.title('Correlation Heatmap (Lower Triangle)', fontsize=16)
plt.tight_layout()
plt.show()

# Compute VIF
X = df_selected.copy()
vif_data = pd.DataFrame()
vif_data['Feature'] = X.columns
vif_data['VIF'] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]

# Display VIF results
print("Variance Inflation Factor (VIF):")
print(vif_data)
````

     - `colsample_bytree`: Fraction of features used for each tree.
   - **Limitations**:  
     - Can be computationally expensive and require careful hyperparameter tuning.

#### **Train 4 Models without Hyperparameter Tuning**

````python
from sklearn.ensemble import GradientBoostingRegressor, AdaBoostRegressor
from xgboost import XGBRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import r2_score, mean_squared_error
import numpy as np
import pandas as pd

def train_and_predict_ensemble_simple(X_train, X_test, y_train_boxcox, y_test_boxcox, pt, y_test_original):
    models = {
        "Gradient Boosting": GradientBoostingRegressor(),
        "AdaBoost": AdaBoostRegressor(),
        "XGBoost": XGBRegressor(),
        "K-Nearest Neighbors": KNeighborsRegressor()
    }

    results = []

    for model_name, model in models.items():
        model.fit(X_train, y_train_boxcox.ravel())
        
        # Predict and inverse transform
        y_train_pred = model.predict(X_train)
        y_test_pred = model.predict(X_test)

        y_train_pred_original = pt.inverse_transform(pd.DataFrame(y_train_pred, columns=["price"]))
        y_test_pred_original = pt.inverse_transform(pd.DataFrame(y_test_pred, columns=["price"]))
        y_train_original = pt.inverse_transform(pd.DataFrame(y_train_boxcox, columns=["price"]))

        # Compute R² and RMSE
        train_r2 = r2_score(y_train_original, y_train_pred_original)
        test_r2 = r2_score(y_test_original, y_test_pred_original)
        train_rmse = np.sqrt(mean_squared_error(y_train_original, y_train_pred_original))
        test_rmse = np.sqrt(mean_squared_error(y_test_original, y_test_pred_original))

        results.append([
            model_name,
            round(train_r2, 4), round(test_r2, 4),
            round(train_rmse, 4), round(test_rmse, 4)
        ])

    results_df = pd.DataFrame(results, columns=[
        "Model", "Train R²", "Test R²", "Train RMSE", "Test RMSE"
    ])

    return results_df
````

````python
results_ensemble_boxcox = train_and_predict_ensemble_simple(X_train, X_test, y_train_boxcox, y_test_boxcox, pt, y_test_original)
````

#### **Train Models with Hyperparameter Tuning (GridSearchCV)**

````python
# Re-import necessary packages after code reset
from sklearn.ensemble import GradientBoostingRegressor, AdaBoostRegressor
from xgboost import XGBRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import GridSearchCV, cross_val_score, learning_curve
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np


# Feature names
features_name = [
    'cut', 'color', 'clarity', 'depth', 'table',
    'PCA_1', 'carat', 'carat^2', 'carat_clarity', 'cut_color'
]

# Define models and grids
models_params = {
    "Gradient Boosting": {
        "model": GradientBoostingRegressor(random_state=42),
        "params": {
            'learning_rate': [0.01, 0.1],
            'n_estimators': [100, 200],
            'max_depth': [3, 5],
            'subsample': [0.8, 1.0]
        }
    },
    "AdaBoost": {
        "model": AdaBoostRegressor(random_state=42),
        "params": {
            'n_estimators': [50, 100],
            'learning_rate': [0.01, 0.1, 1.0]
        }
    },
    "XGBoost": {
        "model": XGBRegressor(random_state=42, verbosity=0, n_jobs=-1),
        "params": {
            'n_estimators': [100, 200],
            'max_depth': [3, 6],
            'learning_rate': [0.01, 0.1],
            'subsample': [0.8, 1],
            'colsample_bytree': [0.8, 1]
        }
    },
    "K-Nearest Neighbors": {
        "model": KNeighborsRegressor(),
        "params": {
            'n_neighbors': [3, 5],
            'weights': ['uniform', 'distance'],
            'p': [1, 2]
        }
    }
}

    
````

#### **Retrieve Best Parameters**

````python
# Train and print best estimators one by one
tuned_models_output = {}

for name, content in models_params.items():
    print("\n" + "=" * 40)
    print(f" TUNED MODEL: {name} ".center(40, "="))
    print("=" * 40)
    
    grid_search = GridSearchCV(
        estimator=content["model"],
        param_grid=content["params"],
        cv=5,
        scoring='r2',
        verbose=0,
        n_jobs=-1,
        refit=True,
        return_train_score=True,
        error_score='raise'
    )
    
    grid_search.fit(X_train, y_train_boxcox.ravel())
    tuned_model = grid_search.best_estimator_
    
    # Store model
    tuned_models_output[name] = tuned_model
    
    # Print tuned model separately
    print(tuned_model)
````

#### **Train Model with Best Parameters**

````python
# Perform predictions and print metrics separately
for name, model in tuned_models_output.items():
    print("\n" + "=" * 40)
    print(f" EVALUATION METRICS: {name} ".center(40, "="))
    print("=" * 40)

    y_pred_tuned = model.predict(X_test)
    y_pred_tuned_original = pt.inverse_transform(pd.DataFrame(y_pred_tuned, columns=["price"]))

    mse = mean_squared_error(y_test_original, y_pred_tuned_original)
    r2 = r2_score(y_test_original, y_pred_tuned_original)

    print(f"Mean Squared Error (MSE): {round(mse, 4)}")
    print(f"R-squared (R²): {round(r2, 4)}")
````

#### **Model Diagnostics (Tuned Model)**

````python
from sklearn.model_selection import cross_val_score, learning_curve
import matplotlib.pyplot as plt
import numpy as np

for name in ["Gradient Boosting", "AdaBoost", "XGBoost", "K-Nearest Neighbors"]:
    model = tuned_models_output[name]

    # Cross-validation
    cv_scores = cross_val_score(
        model, X_train, y_train_boxcox.ravel(), cv=5, scoring='r2'
    )

    # Learning curve
    train_sizes, train_scores, val_scores = learning_curve(
        model,
        X_train,
        y_train_boxcox.ravel(),
        cv=5,
        train_sizes=np.linspace(0.1, 1.0, 10),
        scoring='r2',
        n_jobs=-1
    )

    # Start Plot
    plt.figure(figsize=(18, 6))
    plt.suptitle(f"{name} Diagnostics", fontsize=16)

    # 1. Cross-validation R²
    plt.subplot(1, 2 if name == "K-Nearest Neighbors" else 3, 1)
    plt.bar(range(1, 6), cv_scores, color='#17becf')
    plt.axhline(np.mean(cv_scores), color='red', linestyle='--')
    plt.title(f"5-Fold CV R²\n(Mean: {np.mean(cv_scores):.3f})")
    plt.xlabel("Fold")
    plt.ylabel("R² Score")
    plt.ylim(0, 1)
    plt.grid(axis='y', alpha=0.3)

    # 2. Learning Curve
    plt.subplot(1, 2 if name == "K-Nearest Neighbors" else 3, 2)
    plt.plot(train_sizes, np.mean(train_scores, axis=1), 'o-', label='Training', color='#17becf')
    plt.plot(train_sizes, np.mean(val_scores, axis=1), 'o-', label='Validation', color='#e377c2')
    plt.title("Learning Curve")
    plt.xlabel("Training Samples")
    plt.ylabel("R² Score")
    plt.legend()
    plt.grid(alpha=0.3)

    # 3. Feature Importance (only if applicable)
    if name != "K-Nearest Neighbors":
        importances = model.feature_importances_
        sorted_idx = importances.argsort()

        plt.subplot(1, 3, 3)
        plt.barh(np.array(features_name)[sorted_idx], importances[sorted_idx], color='#17becf')
        plt.title("Feature Importance")
        plt.xlabel("Importance Score")
        plt.grid(axis='x', alpha=0.3)

    plt.tight_layout()
    plt.show()

````
